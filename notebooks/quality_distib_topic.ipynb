```python
import pandas as pd
```

```python
datasets = ["llama3_1_8b", "openbiollm_8b", "gemma2_9b", "medGemma_4b", "medGemma_27b", "qwen3_6b", "mistral_7b", "eurollm_9b", "apertus_8B"]
data_dir = "../data/dataset_with_quality/instruct/"
```

```python
items = pd.read_csv("../data/item.csv")
```

```python
with pd.ExcelWriter('../data/output/output.xlsx') as writer:
    for dataset in datasets:
        df = pd.read_csv(data_dir + dataset + ".csv")
        df["clean-id"] = df["id"].str.slice(0,7)
        result = pd.merge(df, items, left_on="clean-id", right_on="id")
        distribution = result.groupby(["item","distractor_quality"]).size().unstack(fill_value=0)
        #distribution = distribution.round(1)
        distribution['Somme'] = distribution.sum(axis=1).astype(int)
        distribution.to_excel(writer, sheet_name=dataset)
```

```python
clean_ids_set = None
for dataset in datasets:
    df = pd.read_csv(data_dir + dataset + ".csv")
    df["clean-id"] = df["id"].str.slice(0,7)
    dataset_ids = set(df["clean-id"])
    
    if clean_ids_set is None:
        clean_ids_set = dataset_ids
    else:
        clean_ids_set = clean_ids_set.intersection(dataset_ids)

print(clean_ids_set)
```

```python
for i in range(360):
    check = "OIC-"+str(i).zfill(3)  # Ajoute des zéros à gauche
    if check not in clean_ids_set:
        print(check)
```
