```python
import pandas as pd
```

```python
uness = pd.read_json("../data/mcqs_answerability.json")
```

```python
corr = {
    'A' : 1,
    'B' : 2,
    'C' : 3,
    'D' : 4,
}

for idx in range(len(uness)):
    cop = uness["cop"].loc[idx]
    cop = corr[cop]
    uness.loc[idx,"cop"] = cop
    
```

```python
uness.to_json("../data/mcqs_answerability_.json",orient="records",indent=2)
```
